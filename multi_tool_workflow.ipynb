# Multi-Tool Workflow

This notebook demonstrates a workflow using two different tools:
1. **Add Numbers Tool** - A tool for adding two numbers together
2. **Fetch Message Tool** - A tool for fetching a predefined message

## Setup

First, let's import the necessary libraries and define our tools.

```python
from typing import Dict, Any
import json
```

## Tool 1: Add Numbers

This tool takes two numbers as input and returns their sum.

```python
def add_numbers(a: float, b: float) -> Dict[str, Any]:
    """
    Add two numbers together.
    
    Args:
        a: First number
        b: Second number
    
    Returns:
        Dictionary containing the result and operation details
    """
    result = a + b
    return {
        "tool": "add_numbers",
        "inputs": {"a": a, "b": b},
        "result": result,
        "status": "success"
    }
```

## Tool 2: Fetch Message

This tool fetches a message based on a message type.

```python
def fetch_message(message_type: str = "greeting") -> Dict[str, Any]:
    """
    Fetch a predefined message based on the message type.
    
    Args:
        message_type: Type of message to fetch (greeting, farewell, info)
    
    Returns:
        Dictionary containing the message and operation details
    """
    messages = {
        "greeting": "Hello! Welcome to the multi-tool workflow!",
        "farewell": "Goodbye! Thank you for using the workflow.",
        "info": "This is a demonstration of a multi-tool workflow system.",
        "success": "Operation completed successfully!"
    }
    
    message = messages.get(message_type, "Unknown message type")
    
    return {
        "tool": "fetch_message",
        "inputs": {"message_type": message_type},
        "message": message,
        "status": "success"
    }
```

## Workflow Execution

Now let's create a workflow that uses both tools in sequence.

```python
def execute_workflow(num1: float, num2: float, msg_type: str = "greeting") -> Dict[str, Any]:
    """
    Execute a workflow that uses both tools.
    
    Args:
        num1: First number to add
        num2: Second number to add
        msg_type: Type of message to fetch
    
    Returns:
        Dictionary containing results from both tools
    """
    workflow_results = {
        "workflow": "multi_tool_workflow",
        "steps": []
    }
    
    # Step 1: Add numbers
    print("Step 1: Adding numbers...")
    add_result = add_numbers(num1, num2)
    workflow_results["steps"].append(add_result)
    print(f"Result: {add_result['result']}")
    
    # Step 2: Fetch message
    print("\nStep 2: Fetching message...")
    message_result = fetch_message(msg_type)
    workflow_results["steps"].append(message_result)
    print(f"Message: {message_result['message']}")
    
    workflow_results["status"] = "completed"
    return workflow_results
```

## Example 1: Basic Workflow

Let's run the workflow with some sample inputs.

```python
# Execute workflow with sample data
result = execute_workflow(15, 27, "greeting")

# Display the full workflow result
print("\n" + "="*50)
print("WORKFLOW SUMMARY")
print("="*50)
print(json.dumps(result, indent=2))
```

## Example 2: Different Message Types

Let's test different message types while performing different calculations.

```python
# Test with info message
print("Test 1: Info Message")
print("-" * 50)
result1 = execute_workflow(100, 200, "info")

print("\n" + "=" * 50)
print("\nTest 2: Success Message")
print("-" * 50)
result2 = execute_workflow(42, 58, "success")

print("\n" + "=" * 50)
print("\nTest 3: Farewell Message")
print("-" * 50)
result3 = execute_workflow(3.14, 2.86, "farewell")
```

## Individual Tool Testing

You can also use each tool independently.

```python
# Test add_numbers tool individually
print("Testing Add Numbers Tool:")
add_result = add_numbers(50, 75)
print(json.dumps(add_result, indent=2))
```

```python
# Test fetch_message tool individually
print("Testing Fetch Message Tool:")
for msg_type in ["greeting", "farewell", "info", "success"]:
    msg_result = fetch_message(msg_type)
    print(f"\n{msg_type.upper()}: {msg_result['message']}")
```
