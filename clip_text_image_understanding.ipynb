# Task 3: Explore CLIP for Text-to-Image Understanding

**Objective:** Use CLIP to encode text and find the most relevant image from a set of images.

CLIP (Contrastive Language-Image Pre-training) is a neural network trained on a variety of (image, text) pairs. It can be used to find the most relevant image for a given text query by computing similarity scores between text and image embeddings.

## Import Libraries

```python
import torch
from PIL import Image
from transformers import CLIPProcessor, CLIPModel
```

## Load CLIP Model and Processor

We'll use the `openai/clip-vit-base-patch32` model, which is a Vision Transformer (ViT) based CLIP model.

```python
# Load CLIP model and processor
model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32")
processor = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
```

## Load Images and Define Text Query

Load a set of images and define the text query that we want to match against the images.

**Note:** Make sure to replace `"image1.jpg"` and `"image2.jpg"` with paths to your actual image files.

```python
# Images and text
images = [Image.open("image1.jpg"), Image.open("image2.jpg")]
text = ["A serene lake with mountains"]
```

## Encode Images and Text

Use the CLIP processor to prepare the inputs and pass them through the model to get embeddings.

```python
# Encode images and text
inputs = processor(text=text, images=images, return_tensors="pt", padding=True)
outputs = model(**inputs)
```

## Find the Best Matching Image

The CLIP model outputs logits that represent the similarity between the text and each image. We use `argmax` to find the image with the highest similarity score.

```python
# Find the best matching image
logits_per_image = outputs.logits_per_image
best_match = torch.argmax(logits_per_image).item()
print(f"The best matching image is image {best_match + 1}.")
```

## Display Results with Similarity Scores

Let's also display the similarity scores for all images to better understand the model's predictions.

```python
# Display similarity scores for all images
probs = logits_per_image.softmax(dim=0)
print("\nSimilarity scores (probabilities):")
for i, prob in enumerate(probs):
    print(f"  Image {i + 1}: {prob.item():.4f}")
```
