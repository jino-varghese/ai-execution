# String Manipulation Agent

This notebook contains an agent that can perform string manipulation operations based on user queries.

**Supported Operations:**
- **Reverse**: Reverses the input string
- **Capitalize**: Capitalizes the input string

The agent intelligently determines which operation to perform based on keywords in the query.

```python
class StringManipulationAgent:
    """
    An agent that performs string manipulation operations based on user queries.
    """
    
    def __init__(self):
        self.operations = {
            'reverse': self.reverse_string,
            'capitalize': self.capitalize_string,
            'upper': self.capitalize_string,
            'caps': self.capitalize_string,
            'uppercase': self.capitalize_string
        }
    
    def reverse_string(self, text):
        """
        Reverse the given string.
        
        Args:
            text (str): The string to reverse
            
        Returns:
            str: The reversed string
        """
        return text[::-1]
    
    def capitalize_string(self, text):
        """
        Capitalize the given string.
        
        Args:
            text (str): The string to capitalize
            
        Returns:
            str: The capitalized string
        """
        return text.upper()
    
    def process_query(self, query, text):
        """
        Process a user query and apply the appropriate string manipulation.
        
        Args:
            query (str): The query describing the desired operation
            text (str): The text to manipulate
            
        Returns:
            str: The manipulated string
        """
        query_lower = query.lower()
        
        # Determine which operation to perform based on keywords
        for keyword, operation in self.operations.items():
            if keyword in query_lower:
                return operation(text)
        
        # Default to returning the original text if no operation is detected
        return f"No operation detected. Available operations: {', '.join(set(self.operations.keys()))}"
    
    def execute(self, query, text):
        """
        Execute the agent with a query and text.
        
        Args:
            query (str): The query describing the desired operation
            text (str): The text to manipulate
            
        Returns:
            dict: A dictionary containing the original text, query, and result
        """
        result = self.process_query(query, text)
        
        return {
            'query': query,
            'original_text': text,
            'result': result
        }
```

## Initialize the Agent

Create an instance of the StringManipulationAgent.

```python
# Create an instance of the agent
agent = StringManipulationAgent()
print("String Manipulation Agent initialized successfully!")
```

## Example 1: Reverse a String

```python
# Example: Reverse a string
result = agent.execute("Please reverse this text", "Hello World")

print(f"Query: {result['query']}")
print(f"Original Text: {result['original_text']}")
print(f"Result: {result['result']}")
```

## Example 2: Capitalize a String

```python
# Example: Capitalize a string
result = agent.execute("Capitalize this text", "hello world")

print(f"Query: {result['query']}")
print(f"Original Text: {result['original_text']}")
print(f"Result: {result['result']}")
```
