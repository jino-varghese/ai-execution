# Task 5: Visualize a Diffusion Process

This notebook simulates a diffusion process by iteratively adding and removing noise from an image.

Diffusion models work by:
1. **Forward Process**: Gradually adding noise to an image over multiple steps
2. **Reverse Process**: Learning to denoise the image step by step

This simulation demonstrates the forward diffusion process and a simple denoising (clipping) as a placeholder for a trained diffusion model.

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

## Generate a Sample Image

Since we don't have an external image file, we'll create a simple geometric pattern as our sample image.

```python
# Create a sample image (geometric pattern)
def create_sample_image(size=256):
    """Create a sample grayscale image with geometric patterns."""
    image = np.zeros((size, size))
    
    # Add a circle
    center = size // 2
    radius = size // 4
    y, x = np.ogrid[:size, :size]
    mask = (x - center) ** 2 + (y - center) ** 2 <= radius ** 2
    image[mask] = 0.8
    
    # Add a rectangle
    image[size//6:size//3, size//6:size//3] = 0.6
    
    # Add gradient background
    gradient = np.linspace(0.1, 0.3, size).reshape(1, -1)
    image = np.maximum(image, np.tile(gradient, (size, 1)))
    
    return image

# Generate the sample image
image = create_sample_image(256)

# Display the original image
plt.figure(figsize=(6, 6))
plt.title("Original Image")
plt.imshow(image, cmap="gray")
plt.colorbar()
plt.axis('off')
plt.show()
```

## Simulate the Forward Diffusion Process

In the forward diffusion process, we progressively add Gaussian noise to the image. At each step `i`, we add noise with standard deviation proportional to `i`.

```python
# Simulate diffusion (forward process - adding noise)
steps = 10
noise_scale = 0.1  # Base noise level

noisy_images = [image + np.random.normal(0, noise_scale * i, image.shape) for i in range(steps)]

print(f"Generated {steps} noisy images with increasing noise levels")
print(f"Noise standard deviation ranges from 0 to {noise_scale * (steps-1):.2f}")
```

## Denoise Images (Placeholder)

In a real diffusion model, this step would use a trained neural network to predict and remove the noise. Here, we use simple clipping as a placeholder to ensure pixel values remain in the valid range [0, 1].

```python
# Denoise images (placeholder for a diffusion model)
# In practice, this would be a trained neural network that predicts and removes noise
denoised_images = [np.clip(img, 0, 1) for img in noisy_images]

print(f"Denoised {len(denoised_images)} images using clipping")
```

## Visualize the Diffusion Process

Let's visualize how noise affects the image at different steps of the diffusion process.

```python
# Display the diffusion process progression
fig, axes = plt.subplots(2, 5, figsize=(15, 6))
fig.suptitle("Forward Diffusion Process: Adding Noise Over Time", fontsize=14)

for i, ax in enumerate(axes.flatten()):
    ax.imshow(noisy_images[i], cmap="gray", vmin=0, vmax=1)
    ax.set_title(f"Step {i}\nσ = {noise_scale * i:.2f}")
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## Compare Original, Noisy, and Denoised Images

```python
# Display original and final images side by side
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(image, cmap="gray")
axes[0].set_title("Original Image", fontsize=12)
axes[0].axis('off')

axes[1].imshow(noisy_images[-1], cmap="gray")
axes[1].set_title(f"Noisy Image (Step {steps-1})", fontsize=12)
axes[1].axis('off')

axes[2].imshow(denoised_images[-1], cmap="gray")
axes[2].set_title("Denoised Image (Clipped)", fontsize=12)
axes[2].axis('off')

plt.tight_layout()
plt.show()
```

## Visualize Noise Distribution

```python
# Visualize the noise added at different steps
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

steps_to_show = [1, 5, 9]
for ax, step in zip(axes, steps_to_show):
    noise = noisy_images[step] - image
    ax.hist(noise.flatten(), bins=50, density=True, alpha=0.7, color='blue')
    ax.set_title(f"Noise Distribution at Step {step}\nσ = {noise_scale * step:.2f}")
    ax.set_xlabel("Noise Value")
    ax.set_ylabel("Density")

plt.tight_layout()
plt.show()
```

## Summary

This notebook demonstrated:

1. **Forward Diffusion**: How noise is progressively added to an image over multiple steps
2. **Noise Visualization**: How the noise distribution changes at different steps
3. **Simple Denoising**: Using clipping as a placeholder for actual diffusion model denoising

In a real diffusion model (like DDPM, Stable Diffusion, etc.):
- The forward process follows a predefined noise schedule
- The reverse process uses a trained U-Net or Transformer to predict and remove noise
- The model learns to generate realistic images by learning the reverse diffusion process
